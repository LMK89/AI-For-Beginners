# 🧠 Bài Tập Thực Hành Lập Trình Mạng Neural Đơn Giản

Chào mừng bạn đến với bài tập thực hành lập trình mạng Neural. Trong bài tập này, chúng ta sẽ tự viết quá trình lan truyền xuôi (Forward Pass) và lan truyền ngược (Backward Pass) để tối ưu hóa trọng số của một Neuron cơ bản bằng Python và Numpy.

## 📐 1. Định nghĩa các hàm kích hoạt phi tuyến

Chúng ta sẽ định nghĩa hàm ReLU và đạo hàm của nó.

```python
import numpy as np

def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return (x > 0).astype(float)
```

## 🤝 2. Thiết lập nốt Neuron đơn giản

Chúng ta sẽ khởi tạo đầu vào, trọng số và bias mẫu.

```python
inputs = np.array([2.0, 3.0])
weights = np.array([0.5, -0.2])
bias = 0.1

print("Đầu vào inputs:", inputs)
print("Trọng số weights ban đầu:", weights)
print("Bias ban đầu:", bias)
```

## 💻 3. Thực hiện Forward Pass & Backward Pass

Hãy tự điền code tính toán z và đầu ra tương ứng qua hàm ReLU.

```python
# 1. Forward Pass
z = np.dot(inputs, weights) + bias
output = relu(z)
print("Kết quả z =", z)
print("Đầu ra output =", output)

# 2. Giả sử sai số đạo hàm từ lớp sau truyền về là d_loss = 1.0
d_loss = 1.0
d_z = d_loss * relu_derivative(z)

# 3. Tính Gradients cho weights và bias
d_weights = d_z * inputs
d_bias = d_z
print("Độ dốc d_weights =", d_weights)
print("Độ dốc d_bias =", d_bias)

# 4. Cập nhật tham số
learning_rate = 0.1
weights -= learning_rate * d_weights
bias -= learning_rate * d_bias

print("Trọng số weights sau khi cập nhật:", weights)
print("Bias sau khi cập nhật:", bias)
```
