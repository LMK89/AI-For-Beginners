# 🌲 Bài Tập Thực Hành XGBoost & LightGBM

Chào mừng bạn đến với bài tập thực hành về các mô hình dựa trên Cây Quyết Định (Tree-based models). Trong bài tập này, chúng ta sẽ làm quen với việc sử dụng thư viện `lightgbm` và `xgboost` để huấn luyện trên một tập dữ liệu dạng bảng giả lập, đánh giá hiệu năng và vẽ biểu đồ Độ quan trọng của đặc trưng (Feature Importance).

## 📐 1. Chuẩn bị thư viện

Hãy import các thư viện cần thiết cho bài học.

```python
import numpy as np
import pandas as pd
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
import lightgbm as lgb
import xgboost as xgb
```

## 🤝 2. Tạo tập dữ liệu dạng bảng giả lập

Chúng ta sẽ tạo ra một tập dữ liệu giả lập có 1000 hàng và 10 cột đặc trưng biểu diễn cho bài toán phân loại nhị phân (Ví dụ: Dự đoán khách hàng nợ xấu).

```python
X, y = make_classification(n_samples=1000, n_features=10, n_informative=8, n_redundant=2, random_state=42)
feature_names = [f"Đặc trưng_{i}" for i in range(10)]
df = pd.DataFrame(X, columns=feature_names)
df['Nhãn_Thực_Tế'] = y

# Chia tập Train và Test
X_train, X_test, y_train, y_test = train_test_split(df[feature_names], df['Nhãn_Thực_Tế'], test_size=0.2, random_state=42)
print("Kích thước tập Train:", X_train.shape)
print("Kích thước tập Test:", X_test.shape)
df.head()
```

## 💻 3. Huấn luyện mô hình LightGBM

Chúng ta sẽ khởi tạo và cấu hình mô hình LightGBM Classifier.

```python
# Khởi tạo mô hình
model_lgb = lgb.LGBMClassifier(learning_rate=0.05, max_depth=5, n_estimators=100, random_state=42)

# Huấn luyện mô hình
model_lgb.fit(X_train, y_train)

# Dự đoán trên tập Test
y_pred_lgb = model_lgb.predict(X_test)
print("Độ chính xác LightGBM:", accuracy_score(y_test, y_pred_lgb))
print(classification_report(y_test, y_pred_lgb))
```

## 📊 4. Độ quan trọng của Đặc trưng (Feature Importance)

Một trong những ưu điểm lớn nhất của mô hình cây là khả năng giải thích biến số quan trọng nhất.

```python
importances = model_lgb.feature_importances_
importance_df = pd.DataFrame({
    'Đặc Trưng': feature_names,
    'Độ Quan Trọng': importances
}).sort_values(by='Độ Quan Trọng', ascending=False)

print("Bảng xếp hạng đặc trưng quan trọng nhất:")
print(importance_df)
```
