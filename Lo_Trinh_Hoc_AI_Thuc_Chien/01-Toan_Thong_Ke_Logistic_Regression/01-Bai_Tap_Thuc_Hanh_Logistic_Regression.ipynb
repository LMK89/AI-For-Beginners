# 🧠 Bài Tập Thực Hành Lập Trình Hồi Quy Logistic (Logistic Regression Scratch)

Chào mừng bạn đến với bài tập thực hành lập trình từ số không! Trong bài tập này, chúng ta sẽ tự mình xây dựng thuật toán **Hồi Quy Logistic (Logistic Regression)** bằng thư viện Python cơ bản `numpy` để hiểu cặn kẽ từng bước tính toán lý thuyết: Hàm Sigmoid, Hàm Loss Binary Cross-Entropy và tối ưu hóa bằng Gradient Descent.

## 📐 1. Định nghĩa Hàm kích hoạt Sigmoid

Hàm Sigmoid ánh xạ các giá trị số thực bất kỳ thành giá trị xác suất từ 0 đến 1:
$$\sigma(z) = \frac{1}{1 + e^{-z}}$$

```python
import numpy as np

def sigmoid(z):
    """
    Tính toán hàm kích hoạt Sigmoid.
    """
    # YOUR CODE HERE
    return 1 / (1 + np.exp(-z))

# Chạy kiểm thử hàm
print("sigmoid(0) =", sigmoid(0))  # Kết quả mong đợi: 0.5
print("sigmoid(10) =", sigmoid(10))  # Kết quả gần bằng 1
print("sigmoid(-10) =", sigmoid(-10))  # Kết quả gần bằng 0
```

## 🤝 2. Tính toán hàm mất mát Binary Cross-Entropy Loss

Hàm mất mát BCE phạt mô hình dựa trên độ sai lệch xác suất:
$$J = -\frac{1}{m} \sum_{i=1}^{m} [y^{(i)} \log(\hat{y}^{(i)}) + (1 - y^{(i)}) \log(1 - \hat{y}^{(i)})]$$

```python
def compute_loss(y_true, y_pred):
    """
    Tính toán Binary Cross-Entropy Loss.
    Cần cộng thêm 1e-15 để tránh lỗi log(0).
    """
    m = len(y_true)
    eps = 1e-15
    # YOUR CODE HERE
    loss = - (1/m) * np.sum(y_true * np.log(y_pred + eps) + (1 - y_true) * np.log(1 - y_pred + eps))
    return loss

# Chạy kiểm thử hàm
y_t = np.array([1, 0, 1])
y_p = np.array([0.9, 0.1, 0.95])
print("BCE Loss =", compute_loss(y_t, y_p))  # Kết quả mong đợi: Rất nhỏ
```

## 💻 3. Lập trình Class Logistic Regression hoàn chỉnh

Chúng ta sẽ đóng gói toàn bộ thuật toán vào Class hướng đối tượng.

```python
class LogisticRegressionScratch:
    def __init__(self, lr=0.1, epochs=1000):
        self.lr = lr
        self.epochs = epochs
        self.weights = None
        self.bias = None
        self.loss_history = []
        
    def fit(self, X, y):
        m, n = X.shape
        self.weights = np.zeros(n)
        self.bias = 0.0
        
        for epoch in range(self.epochs):
            # Forward Pass
            z = np.dot(X, self.weights) + self.bias
            y_pred = sigmoid(z)
            
            # Compute Gradients
            dw = (1/m) * np.dot(X.T, (y_pred - y))
            db = (1/m) * np.sum(y_pred - y)
            
            # Update Parameters
            self.weights -= self.lr * dw
            self.bias -= self.lr * db
            
            # Save Loss History
            loss = compute_loss(y, y_pred)
            self.loss_history.append(loss)
            
    def predict_proba(self, X):
        z = np.dot(X, self.weights) + self.bias
        return sigmoid(z)
        
    def predict(self, X, threshold=0.5):
        return (self.predict_proba(X) >= threshold).astype(int)
```

## 📊 4. Chạy thực nghiệm

Hãy thử nghiệm trên một bộ dữ liệu giả lập đơn giản.

```python
# Tạo dữ liệu mẫu
X_sample = np.array([[1.0, 2.0], [2.0, 3.0], [3.0, 1.0], [5.0, 5.0], [6.0, 7.0], [7.0, 6.0]])
y_sample = np.array([0, 0, 0, 1, 1, 1])

# Huấn luyện mô hình
model = LogisticRegressionScratch(lr=0.1, epochs=500)
model.fit(X_sample, y_sample)

print("Trọng số weights học được:", model.weights)
print("Bias học được:", model.bias)

# Dự đoán thử mẫu mới
new_point = np.array([[1.5, 1.5], [6.5, 6.5]])
print("Xác suất dự đoán cho mẫu mới:", model.predict_proba(new_point))
print("Nhãn dự đoán cho mẫu mới:", model.predict(new_point))
```
