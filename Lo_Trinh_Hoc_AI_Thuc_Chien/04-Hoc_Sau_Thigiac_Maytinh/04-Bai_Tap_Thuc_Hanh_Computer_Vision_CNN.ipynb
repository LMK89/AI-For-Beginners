# 👁️ Bài Tập Thực Hành Học Sâu Cho Thị Giác Máy Tính (CNN bằng PyTorch)

Chào mừng bạn đến với bài tập thực hành xây dựng mạng tích chập CNN bằng PyTorch. Ở bài tập này, chúng ta sẽ định nghĩa một lớp mạng CNN chuẩn, tìm hiểu cách hoạt động của lớp tích chập Conv2D và lớp MaxPool2D trên ma trận ảnh giả lập.

## 📐 1. Chuẩn bị thư viện

Hãy import thư viện `torch` và `torch.nn`.

```python
import torch
import torch.nn as nn
```

## 🤝 2. Định nghĩa cấu trúc mạng SimpleCNN

Chúng ta sẽ định nghĩa cấu trúc lớp mạng CNN nhận diện ảnh đơn sắc.

```python
class SimpleCNN(nn.Module):
    def __init__(self, num_classes=10):
        super(SimpleCNN, self).__init__()
        # Conv2D: Nhận ảnh 1 kênh (Grayscale), xuất ra 8 kênh đặc trưng, kích thước bộ lọc 3x3
        self.conv = nn.Conv2d(in_channels=1, out_channels=8, kernel_size=3, padding=1)
        self.relu = nn.ReLU()
        # MaxPool2D: Thu nhỏ kích thước ma trận đi một nửa
        self.pool = nn.MaxPool2d(kernel_size=2, stride=2)
        # Fully Connected Layer để phân loại lớp học
        self.fc = nn.Linear(8 * 14 * 14, num_classes)
        
    def forward(self, x):
        x = self.conv(x)
        x = self.relu(x)
        x = self.pool(x)
        # Phẳng hóa ma trận đặc trưng
        x = x.view(x.size(0), -1)
        x = self.fc(x)
        return x

# Khởi tạo mô hình mạng
model = SimpleCNN()
print(model)
```

## 💻 3. Chạy thử nghiệm trên ma trận ảnh giả lập

Chúng ta tạo ra một Tensor có dạng Batch x Channels x Height x Width mô phỏng ảnh kích thước 28x28.

```python
# Tạo Tensor ảnh giả lập (1 ảnh, 1 kênh màu, kích thước 28x28)
fake_image = torch.randn(1, 1, 28, 28)

# Đưa ảnh qua mô hình
predictions = model(fake_image)
print("Kích thước đầu ra dự đoán (Batch size, Num classes) =", predictions.shape)
print("Xác suất dự đoán cho 10 lớp học =", predictions.detach().numpy())
```
