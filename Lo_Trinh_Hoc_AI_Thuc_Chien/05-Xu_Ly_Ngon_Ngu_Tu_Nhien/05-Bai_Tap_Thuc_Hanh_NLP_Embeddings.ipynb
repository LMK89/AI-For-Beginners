# 🗣️ Bài Tập Thực Hành Xử Lý Ngôn Ngữ Tự Nhiên (Tính Toán Vector Embedding)

Chào mừng bạn đến với bài tập thực hành NLP. Trong bài tập này, chúng ta sẽ tự lập trình công thức tính **Độ tương đồng Cosine (Cosine Similarity)** bằng thư viện `numpy` để so khớp khoảng cách hình học ngữ nghĩa giữa hai từ/câu văn.

## 📐 1. Định nghĩa công thức Cosine Similarity

Công thức đo lường độ tương đồng góc giữa 2 vectơ:
$$\text{Similarity} = \frac{\vec{u} \cdot \vec{v}}{\|\vec{u}\| \|\vec{v}\|}$$

```python
import numpy as np

def cosine_similarity(u, v):
    """
    Tính toán độ tương đồng Cosine giữa 2 vectơ u và v.
    """
    # YOUR CODE HERE
    dot_product = np.dot(u, v)
    norm_u = np.linalg.norm(u)
    norm_v = np.linalg.norm(v)
    if norm_u == 0 or norm_v == 0:
        return 0.0
    return dot_product / (norm_u * norm_v)
```

## 🤝 2. Chạy thử nghiệm trên các vectơ từ khóa mẫu

Giả sử chúng ta có các vectơ đặc trưng 4 chiều biểu thị cho các từ khóa.

```python
# Khởi tạo vectơ đặc trưng giả lập
v_hoc_may = np.array([0.9, 0.8, 0.1, 0.0])
v_tri_tue = np.array([0.85, 0.75, 0.15, 0.0])
v_thoi_tiet = np.array([0.1, 0.0, 0.9, 0.95])

# So sánh
sim_chuyen_nganh = cosine_similarity(v_hoc_may, v_tri_tue)
sim_khac_biet = cosine_similarity(v_hoc_may, v_thoi_tiet)

print(f"Độ tương đồng giữa 'học máy' và 'trí tuệ nhân tạo' (cùng chuyên ngành) = {sim_chuyen_nganh:.4f}")
print(f"Độ tương đồng giữa 'học máy' và 'thời tiết' (khác chủ đề) = {sim_khac_biet:.4f}")
```
