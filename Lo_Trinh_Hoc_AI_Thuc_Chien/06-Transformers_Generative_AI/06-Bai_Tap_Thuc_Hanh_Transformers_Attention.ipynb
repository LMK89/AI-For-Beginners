# 🤖 Bài Tập Thực Hành Lập Trình Cơ Chế Chú Ý (Self-Attention bằng PyTorch)

Chào mừng bạn đến với bài tập thực hành đỉnh cao! Trong bài tập này, chúng ta sẽ tự tay lập trình cơ chế toán học cốt lõi đứng sau ChatGPT - cơ chế **Scaled Dot-Product Attention** bằng PyTorch từ số không.

## 📐 1. Chuẩn bị thư viện

Import các thư viện chuyên dụng.

```python
import torch
import torch.nn.functional as F
import math
```

## 🤝 2. Lập trình hàm Scaled Dot-Product Attention

Công thức cốt lõi:
$$\text{Attention}(Q, K, V) = \text{Softmax}\left(\frac{Q K^T}{\sqrt{d_k}}\right) V$$

```python
def scaled_dot_product_attention(Q, K, V):
    """
    Tính toán ma trận kết quả Attention.
    """
    d_k = Q.size(-1)
    
    # 1. Nhân ma trận đo độ tương đồng Q và K_transpose
    scores = torch.matmul(Q, K.transpose(-2, -1))
    
    # 2. Co giãn điểm số
    scores = scores / math.sqrt(d_k)
    
    # 3. Áp dụng Softmax
    attention_weights = F.softmax(scores, dim=-1)
    
    # 4. Nhân với Value
    output = torch.matmul(attention_weights, V)
    
    return output, attention_weights
```

## 💻 3. Chạy thử nghiệm trên các Tensor giả lập

Chúng ta mô phỏng một câu văn có 3 từ (Sequence length = 3), mỗi từ biểu thị bằng vector 4 chiều (Dimension = 4).

```python
# Tạo các Tensor Q, K, V giả lập kích thước (Batch size=1, Seq length=3, Dim=4)
Q = torch.randn(1, 3, 4)
K = torch.randn(1, 3, 4)
V = torch.randn(1, 3, 4)

# Tính toán kết quả
output, weights = scaled_dot_product_attention(Q, K, V)
print("Ma trận kết quả đầu ra output (Batch size, Seq length, Dim) =", output.shape)
print("Trọng số chú ý Attention Weights (Batch size, Seq length, Seq length) =", weights.shape)
print("Ma trận trọng số chú ý chi tiết:\n", weights.detach().numpy())
```
